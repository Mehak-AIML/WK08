## Read filenames into lists

```python
files = [
  sorted(["./data/image/clouds/"+f for f in listdir("./data/image/clouds") if f.endswith("jpg")]),
  sorted(["./data/image/flowers/"+f for f in listdir("./data/image/flowers") if f.endswith("png")]),
  sorted(["./data/image/metfaces/256/"+f for f in listdir("./data/image/metfaces/256") if f.endswith("jpg")]),
  sorted(["./data/image/bob-ross/"+f for f in listdir("./data/image/bob-ross") if f.endswith("jpg")]),
]

[len(fs) for fs in files]
```

## Get pixels

Shrink images and get a flat list of R,G,B values into a single list for each image.

This can take a while.

We can also filter the list of files and only do $400$ images of each class/category.

```python
data = []

# for each category of image
for class_idx,fs in enumerate(files):
  # for each image in a category
  for f in fs:
    # open and shrink image to 32x32 pixels
    img = PImage.open(f).resize((32,32)).convert("RGB")
    pxs = get_pixels(img)

    # create flat list of pixel channel values (should be 3072 values in the end 32x32x3)
    flat_pxs = []
    for r,g,b in pxs:
      flat_pxs.append(r)
      flat_pxs.append(g)
      flat_pxs.append(b)

    # append the category id in the end of the list
    # 0 for clouds, 1 for flowers, 2 for faces, 3 for bob ross
    flat_pxs.append(class_idx)
    data.append(flat_pxs)
```

## Create `DataFrame`

```python
data_df = pd.DataFrame(data)
class_col = data_df.columns[-1]
data_df.rename({ class_col:"class" }, axis=1, inplace=True)
data_df
```
